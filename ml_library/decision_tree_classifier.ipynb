```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ml_library import DecisionTreeClassifier , train_test_split , z_score_normalization
```

```python
df = pd.read_csv("./dataset/train_binary.csv")
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
X = df.drop("label", axis=1).values
y = df["label"].values.reshape(-1,1)
X.shape, y.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
X_train = z_score_normalization(X_train)
X_test = z_score_normalization(X_test)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
model = DecisionTreeClassifier(min_sample_split=3 , max_depth= 3)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_pred = np.array(y_pred, dtype=int)
y_test = y_test.reshape(-1)
model.evaluation_metrics(y_test , y_pred)
```

```python
df_test = pd.read_csv("./dataset/test_binary.csv")
df_test.shape
```

```python
df_test.sample(5)
```

```python
X_testing = df_test.values
X_testing = z_score_normalization(X_testing)
```

```python
y_pred_test = model.predict(X_testing)
y_pred_test = np.array(y_pred_test, dtype=int)
y_pred_test
```

```python
df_test["label"] = y_pred_test
df_test.head()
```

```python
df_test.to_csv("./output/predictions_decision_tree_classifier.csv", index=False)
```
