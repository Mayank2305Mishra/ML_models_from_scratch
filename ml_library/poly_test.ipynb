```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from ml_library import PolynomialRegression, train_test_split, z_score_normalization
```

```python
df = pd.read_csv("./dataset/Spectacle_Prices_Dataset.csv")

df.shape
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1)
```

```python
X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

```python
X_train = z_score_normalization(X_train)
X_test = z_score_normalization(X_test)
```

```python
alpha = 0.1
iters = 2000
model = PolynomialRegression(degree=3, iters=iters, alpha=alpha)
model.fit(X_train, y_train)
```

```python
model.plot_cost()
```

```python
y_pred = model.predict(X_test)
```

```python
model.evaluation_metrics(y_test, y_pred)
```
